## Skill 2: Counting Category Values with `value_counts()`

This suggests that parking and noise issues are major sources of 311 service request volume in this sample.

This suggests that a large portion of the sampled 311 workload is related to issues routed to NYPD, such as illegal parking and noise complaints.

```python
# Request status distribution
df["status"].value_counts()
```

This status distribution is useful because request status affects how the dataset should be interpreted. For example, missing `closed_date` values may be expected for requests that are still open or in progress.

## Skill 3: Filtering Rows

Filtering rows means keeping only the records that match a specific condition.

For the NYC 311 dataset, filtering is useful when analyzing subsets such as complaints from one borough, complaints handled by one agency, noise-related complaints, or requests with a specific status.

```python
analysis_columns = [
    "created_date",
    "closed_date",
    "agency",
    "agency_name",
    "complaint_type",
    "descriptor",
    "borough",
    "status",
    "city"
]

print("analysis_columns created.")
```

```python
# Filter for Brooklyn complaints
brooklyn_df = df[df["borough"] == "BROOKLYN"]

print("Brooklyn complaints:", brooklyn_df.shape[0])

brooklyn_df[analysis_columns].head()
```

### Observation: Brooklyn Complaints

Filtering for `BROOKLYN` creates a smaller DataFrame containing only Brooklyn service requests. This makes it possible to analyze complaint patterns within one borough instead of the full citywide dataset.

```python
# Filter for noise-related complaints
noise_df = df[df["complaint_type"].str.contains("Noise", case=False, na=False)]

print("Noise-related complaints:", noise_df.shape[0])

noise_df[analysis_columns].head()
```

### Observation: Noise-Related Complaints

There are 24,737 noise-related complaints in this 100,000-row NYC 311 sample.

The filter searched for complaint types containing the word `Noise`, which includes categories such as `Noise - Commercial`, `Noise - Residential`, and `Noise - Street/Sidewalk`.

This subset will be useful for analyzing noise complaint patterns by borough, status and agency.

```python
# Filter for noise-related complaints in Brooklyn
brooklyn_noise_df = df[
    (df["borough"] == "BROOKLYN") &
    (df["complaint_type"].str.contains("Noise", case=False, na=False))
]

print("Brooklyn noise-related complaints:", brooklyn_noise_df.shape[0])

brooklyn_noise_df[analysis_columns].head()
```

### Observation: Brooklyn Noise Complaints

Filtering with two conditions created a subset of records that are both located in `BROOKLYN` and noise-related.

This type of filtering is useful because many business questions are not citywide. Analysts often need to answer questions for a specific location, complaint category, agency or status.

## Skill 4: Sorting Values

Sorting helps arrange records or summary results in a meaningful order.

## Skill 5: Converting Date Columns

The `created_date` and `closed_date` columns were loaded as object/text columns.

To perform reliable time-based analysis, these columns should be converted to datetime format. This allows pandas to correctly identify earliest dates, latest dates, and later calculate resolution time.

```python
# Convert date columns from text/object to datetime
df["created_date"] = pd.to_datetime(df["created_date"], errors="coerce")
df["closed_date"] = pd.to_datetime(df["closed_date"], errors="coerce")

print("Date conversion complete.")
print(df[["created_date", "closed_date"]].dtypes)
```

```python
# Convert date columns from text/object to datetime
df["created_date"] = pd.to_datetime(df["created_date"], errors="coerce")
df["closed_date"] = pd.to_datetime(df["closed_date"], errors="coerce")

print("Date conversion complete.")
print(df[["created_date", "closed_date"]].dtypes)
```

```python
# Date range covered by the dataset
earliest_created_date = df["created_date"].min()
latest_created_date = df["created_date"].max()

print("Earliest created date:", earliest_created_date)
print("Latest created date:", latest_created_date)
```

### Observation: Date Range

The 100,000-row sample covers requests created between June 16, 2026 and June 26, 2026.

This means the notebook is analyzing a recent sample of NYC 311 requests rather than the full historical dataset.
